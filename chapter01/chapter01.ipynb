## 2 - The `TinyAgent`

Let's start with building our `TinyAgent`. This class will start with the LLM we defined above and nothing more. It technically cannot be considered an Agent yet until we have added all above components.

That said, let's build the scaffolding of this `TinyAgent`. There are several components that we can already define without using any LLM, namely:

* Attributes that we add later on
* `run` which is the main function to run the Agent
* `_step` which is a single step of the Agent (it may run for multiple steps)

````python
class TinyAgent:
    """A minimal, modular, and educational agent framework."""

    def __init__(self):
        self.llm = None  # Chapter 2 & 3: Add LLM
        self.memory = None  # Chapter 4: Add Memory
        self.tools = None  # Chapter 5: Add Tools
        self.planner = None  # Chapter 6: Add Planning

    def run(self, task: str) -> str:
        """Run the agent on a task."""
        return self._step(task)

    def _step(self, task: str) -> str:
        """Perform a single step."""
        # Placeholder - will be implemented in later chapters
        return f"Received: {task}"

    def _execute_action(self, action: str) -> str:
        """Execute a tool action."""
        # Placeholder - will be implemented in later chapters
        return f"Executed action: {action}"
````

# What's Next

In the next chapter, we cover how LLMs work, which inference providers you can choose from, and how we are going to use it as the "brain" in your `TinyAgent`.

 In Chapter 1, `TinyAgent` (cell 8 of the notebook) does almost nothing on purpose. It gives the agent its basic shape, and later chapters fill each part in. To see why it has this shape, start with what an agent is.
## First principles: what is an agent?

1. **An LLM on its own is a pure function: text in, text out.** You send a prompt and get one reply back. It doesn't remember anything between calls and can't act on the world.

2. **An agent wraps that function in a loop.** It does more than answer once. It repeats a cycle:
   - **Think:** ask the LLM what to do next.
   - **Act:** if the LLM asks for a tool (calculator, search, file read…), run it.
   - **Observe:** feed the tool's result back to the LLM.
   - **Repeat** until the LLM says "I'm done, here's the answer", or a step limit is hit.

3. **Running that loop takes four ingredients:**

   | Ingredient | Why it's needed |
   |---|---|
   | **LLM** (the "brain") | Decides what to do at each step |
   | **Memory** | The LLM is stateless, so something has to keep the conversation and past observations and resend them on every call |
   | **Tools** | Let the agent do things beyond producing text |
   | **Planner** | Defines the loop's rules (here ReAct: Thought → Action → Observation), how to parse the LLM's output, and when to stop |

## The Chapter 1 class, piece by piece

```python
def __init__(self):
    self.llm = None      # Ch 2–3
    self.memory = None   # Ch 4
    self.tools = None    # Ch 5
    self.planner = None  # Ch 6
```
These are the four ingredients, each set to `None` for now. Each comment says which chapter fills it in.

```python
def run(self, task):
    return self._step(task)
```
`run` is the public entry point: "here's a task, give me the final answer." Eventually it will hold the loop. For now it makes exactly one step.

```python
def _step(self, task):
    return f"Received: {task}"
```
`_step` is one turn of the loop (one think, plus maybe one action). With no LLM yet, it just echoes the input. That's why `agent.run("What is 2 + 2?")` returns `"Received: What is 2 + 2?"`. In the book's words, it has no brain yet.

```python
def _execute_action(self, action):
    return f"Executed action: {action}"
```
This is where tool calls will run. For now nothing calls it.

The point of the chapter is the split between **`run`** (the loop that keeps going until done) and **`_step`** (a single iteration). That split is the core of every agent framework.

## Where it ends up

The finished version in `src/illustrated_agents/agent.py` has the same skeleton with the placeholders filled in:

- **`__init__`** (`agent.py:11-31`): takes real `llm`, `memory`, `tools` and `planner` objects. It builds a system prompt that tells the LLM how to plan and which tools exist, and stores that prompt in memory.
- **`run`** (`agent.py:33-44`): now contains the loop, `for step in range(self.planner.max_steps)`. It calls `_step()` until one returns a non-`None` result, or gives up with "Max steps reached".
- **`_step`** (`agent.py:46-67`):
  1. **Thought:** sends the whole memory to the LLM and stores the reply.
  2. Parses the reply to work out whether it's a tool call or a final answer.
  3. **Answer:** if the agent is done (`tools.is_done`), returns the answer and the loop ends.
  4. Otherwise it calls `_execute_action`.
- **`_execute_action`** (`agent.py:69-82`): **Action**, runs the tool. **Observation**, adds the tool's result to memory so the LLM sees it on the next step. It returns `None`, which tells `run` to keep looping.

## Summary
`TinyAgent` in Chapter 1 is an empty shell with the right shape: four slots for the components every agent needs, and a `run` → `_step` → `_execute_action` call structure for the think–act–observe loop. It's built this way so that in each later chapter you can see exactly which slot you're filling and why the agent needs it.
